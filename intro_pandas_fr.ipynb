# Manipulation des Series

Dans cette section, nous en apprenons plus sur la manipulation des ``Series``.

👉 Un ``Series`` est très similaire à un ``numpy.ndarray``. En fait, un ``Series`` est souvent appelée un vecteur. La plupart des méthodes que nous pouvons utiliser sur un `numpy.ndarray`, peuvent être utilisées sur un `Series` également.


---

**Documentation** Tous les attributs et méthodes d'un ``Séries`` [sont documentés ici](https://pandas.pydata.org/docs/reference/api/pandas.Series.html).

## Attributs d'un Series

Dans la cellule suivante, nous présentons quelques-uns des attributs les plus importants d'un objet ``Series``.

```python
continent_col = df['continent']

print(f"Première valeur dans la colonne : {continent_col.loc[0]}")

print(f"Type des valeurs dans la colonne : {continent_col.dtype}")

print(f"Nombre de valeurs dans la colonne : {continent_col.size}")

print(f"Dimension d'une colonne : {continent_col.shape}")
```

La procédure suivante permet d'obtenir un ``ndarray`` à partir d'un ``Series``.

```python
continents = continent_col.values

print(f"Type de la nouvelle variable : {type(continents)}")

print(f"Valeurs : {continents}")
```

## Méthodes d'un Series

Plusieurs méthodes sont disponibles pour calculer les statistiques de base d'un ``Series``.

```python
life_exp_col = df['lifeExp']

print(f"Espérance de vie maximale : {life_exp_col.max()}")
print(f"Espérance de vie minimale : {life_exp_col.min()}")
print(f"Espérance de vie moyenne : {life_exp_col.mean()}")
print(f"Écart-type de l'espérance de vie : {life_exp_col.std()}")
print(f"Médiane de l'espérance de vie : {life_exp_col.median()}")
```

La méthode ``describe`` peut être utilisée pour obtenir toutes les statistiques importantes sur un ``Series`` en une seule instruction.

```python
print(life_exp_col.describe())
```

Si vous préférez les diagrammes, vous pouvez facilement obtenir un histogramme.

Dans cet exemple, l'affichage a été obtenu grâce à une commande du module ``matplotlib``.

```python
import matplotlib.pyplot as plt

life_exp_col.hist()

plt.show()
```

## Masque booléen

Lorsque nous avons obtenu des sous-ensembles de lignes dans les exemples ci-dessus, nous avons supposé que nous connaissions déjà les index des lignes à sélectionner.

Dans de nombreuses situations, vous voudriez sélectionner des lignes selon qu'elles remplissent ou non une certaine condition.

---

**Exemple** : vous souhaitez obtenir des informations sur les localités où l'espérance de vie est inférieure à la moyenne.

---

Notre point de départ est l'expression booléenne suivante.

👉 L'expression renvoie un autre ``Series``, où chaque élément est un booléen.

```python
life_exp_col < life_exp_col.mean()
```

L'expression compare chaque valeur de ``life_exp_col`` à la valeur moyenne et renvoie ``True`` ou ``False``, selon le résultat de la comparaison.

👉 Bien que cette expression implique des comparaisons multiples, une boucle ``for`` n'est pas nécessaire.
C'est un exemple d'opération **vectorisée**, qui travaille sur un vecteur entier sans avoir besoin d'une boucle explicite.

Nous pouvons maintenant utiliser directement cette expression booléenne pour sélectionner les valeurs de l'espérance de vie qui sont inférieures à la moyenne.

```python
print(life_exp_col[life_exp_col < life_exp_col.mean()])
```

### En savoir plus sur la vectorisation

La vectorisation ne fonctionne pas seulement avec l'opérateur de comparaison.
Le code suivant renvoie une nouvelle ``Series`` qui contient la somme élément par élément du vecteur ``life_exp_col``.

```python
print(life_exp_col + life_exp_col)
```

La vectorisation fonctionne également avec une opération.

```python
print(life_exp_col + 10)
```

## Manipuler des dates

La colonne ``publication_date`` contient des dates.

* Toutes les dates sont au format ``mois/jour/année``.

* Cependant, le format de la date n'est pas cohérent d'une ligne à l'autre (le jour ou le mois peuvent être donnés sur un ou deux chiffres).

* Le type de données de la colonne est ``objet``. Les dates sont encodées sous forme de chaînes de caractères.

```python
books_df.publication_date
```

Vous pouvez utiliser la fonction ``to_datetime`` pour convertir la colonne en un objet ``datetime64``. Cela rendra le format de la date cohérent sur toutes les lignes.

Pandas peut déduire le format des dates. Si certaines dates sont erronées (par exemple, 11/30/1980), la conversion n'est pas terminée et aucune erreur n'est affichée.

```python
books_df.publication_date = pd.to_datetime(books_df.publication_date)

books_df.publication_date
```

Alternativement, vous pouvez convertir n'importe quelle colonne contenant des dates lorsque vous chargez les données à partir du fichier CSV.

👉 La méthode ``read_csv`` a un paramètre ``parse_dates``. [Voir la documentation](https://pandas.pydata.org/docs/reference/api/pandas.read_csv.html) pour plus de détails.

```python
books_df = pd.read_csv("./data/books.csv", delimiter=";", parse_dates=['publication_date'])

books_df.publication_date
```

### Extraction des composants d'une date

Nous pouvons extraire les différents composants d'une date (année, mois et jour) à partir d'un objet ``datetime64``.

Le code dans la cellule suivante crée trois nouvelles colonnes ``pub_day``, ``pub_month`` et ``pub_year``.

👉 Notez l'utilisation de l'accesseur ``dt`` pour accéder aux propriétés de date et d'heure d'un ``Series``.

### Changer le format de la date

Dans notre cours, nous avons choisi de représenter les dates dans une base de données SQLite sous forme de chaînes de caractères.

Par conséquent, avant d'importer les données dans une base de données, nous devons convertir nos dates en ``string`` et, ce faisant, modifier le format de la date.

Le code de la cellule suivante convertit les dates au format JJ/MM/AAAA :

* La fonction ``strftime`` renvoie un nouveau ``Series``, dont chaque élément est une date convertie au format spécifié.

* Le format est donné sous la forme d'une chaîne formatée. Dans notre cas, ``"%d/%m/%Y"`` signifie que chaque date sera au format JJ/MM/AAA.

👉 Vous pouvez [consulter la documentation](https://docs.python.org/3/library/datetime.html#strftime-and-strptime-behavior) pour plus d'options de format.

```python
books_df.publication_date = books_df.publication_date.dt.strftime("%d/%m/%Y")

books_df.publication_date
```

## Vérifier les données manquantes

Nous allons maintenant charger les mêmes données mais avec des données manquantes.

```python
books_df = pd.read_csv("./data/books_with_na.csv", delimiter=";")
```

Nous pouvons sortir le taux de données manquantes par colonne :

```python
mask = books_df.isna()
mask
```

On obtient un masque booléen.

Si on rajoute la somme, on obtient le nombre de valeurs manquantes par colonne :

```python
na_sum = mask.sum()
na_sum
```

On peut maintenant diviser par le nombre de lignes du dataset : ``len(books_df)``
et multiplier par 100 (``*100``) pour obtenir le taux de données manquantes :

```python
na_percentage = round(books_df.isna().sum()/len(books_df)*100, 4)
na_percentage = round(na_sum/len(books_df)*100, 4)
na_percentage
```

On peut à nouveau se servir du masque booléen afin d'isoler les lignes avec des données manquantes :

```python
rows_with_na = books_df[books_df["authors"].isna() | books_df["average_rating"].isna()]
rows_with_na
```

Comment remplacer les données manquantes ?

Pour les auteurs, il suffit de les chercher sur Internet avec le titre du livre.

Une fois obtenus, ``.loc`` est pratique pour modifier des valeurs en fonction de l'index de la ligne :

```python
books_df.loc[[5034, 5086, 5087],"authors"] = "Leo Tolstoy"
books_df.loc[6888:6890,"authors"] = "Honoré de Balzac"
```

Pour les prix, on peut éventuellement remplacer par la moyenne avec la méthode ``fillna``. Même si ce n'est pas la méthode la plus efficace (ici, il faudrait retrouver la source des notes et la remettre comme pour les auteurs), elle a l'avantage d'être neutre pour l'analyse de la variable :
